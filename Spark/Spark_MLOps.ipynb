#### Analyse search data

```python
# How many times was the term `gaming laptop` searched?
# Take a screenshot of the code and name it as gaminglaptop.jpg)
```

```python
# Count how many times "gaming laptop" was searched
gaming_laptop_count = searchterms_data.filter(
    searchterms_data["searchterm"] == "gaming laptop"
).count()

print(f"'gaming laptop' was searched {gaming_laptop_count} times")
```

```python
# Print the top 5 most frequently used search terms?
# Take a screenshot of the code and name it as top5terms.jpg)
```

```python
from pyspark.sql.functions import col, desc
```

```python
# Group by search term, count occurrences, and sort descending
top_5_terms = (
    searchterms_data.groupBy("searchterm")
    .count()
    .orderBy(desc("count"))
    .limit(5)
)

top_5_terms.show()
```

```python
# Using the sales forecast model, predict the sales for the year of 2023.
# Take a screenshot of the code and name it as forecast.jpg
```

```python
def predict(year):
    assembler = VectorAssembler(inputCols=["year"], outputCol="features")
    data = [[year, 0]]
    columns = ["year", "sales"]
    df = spark.createDataFrame(data, columns)
    transformed_df = assembler.transform(df).select('features', 'year')
    predictions = model.transform(transformed_df)
    predictions.select('prediction').show()
```
